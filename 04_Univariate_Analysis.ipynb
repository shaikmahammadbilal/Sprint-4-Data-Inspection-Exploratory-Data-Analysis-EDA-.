# Notebook 4 – Univariate Analysis

**Definition:** Univariate analysis examines one variable at a time to understand its distribution, typical value, spread, and unusual values.

**Simple example:** Calculate the average of five test scores.

**Real-world example:** Study the distribution of patient temperatures.

**Business example:** Analyze the distribution of order values.

**AI/ML use case:** Check feature skew, outliers, and target imbalance before preprocessing.

**Importance:** It helps understand each column before exploring relationships between columns.




## 1. Load the dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.DataFrame({
    "employee_id": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1009, 1010, 1011, 1012],
    "age": [22, 25, 29, 31, 35, 41, 27, 38, 45, 24, np.nan, 31],
    "department": ["IT", "HR", "IT", "Sales", "IT", "Finance", "Sales", "HR", "Finance", "IT", "Sales", "it "],
    "education": ["Bachelor", "Master", "Bachelor", "Bachelor", "Master", "PhD", "Bachelor", "Master", "Bachelor", "Bachelor", "Master", "Bachelor"],
    "monthly_salary": [28000, 35000, 45000, 42000, 65000, 82000, 39000, 56000, 90000, 32000, 47000, 45000],
    "monthly_spending": [12000, 16000, 20000, 19000, 25000, 30000, 18000, 24000, 32000, 15000, 21000, 20000],
    "remote": ["No", "Yes", "Yes", "No", "Yes", "Yes", "No", "Yes", "No", "No", "Yes", "Yes"],
    "joined_date": pd.to_datetime(["2022-01-10", "2021-05-15", "2020-03-01", "2023-02-20", "2018-07-12", "2016-11-30", "2022-09-05", "2019-04-18", "2014-08-25", "2023-06-01", "2020-12-14", "2021-08-09"]),
    "left_company": ["No", "No", "No", "Yes", "No", "No", "Yes", "No", "No", "No", "Yes", "No"]
})
df
```

## 2. Numerical variables: Mean, Median, Mode, Minimum, Maximum, and Range
**Definitions:** Mean is the arithmetic average; median is the middle value; mode is the most frequent value; minimum/maximum are the extremes; range is maximum minus minimum.

**Simple example:** For [2, 3, 3, 10], mean = 4.5, median = 3, mode = 3, range = 8.

**Real-world example:** Median travel time can describe a typical commute when a few journeys are unusually long.

**Business example:** Mean and median order value help understand typical customer purchases.

**AI/ML use case:** These statistics reveal skew and extreme values that may require transformations or robust scaling.

**Importance:** Mean is sensitive to extremes; median is more robust. Compare both rather than relying on one.

```python
values = pd.Series([2, 3, 3, 10])
print("Mean:", values.mean())
print("Median:", values.median())
print("Mode:", values.mode().tolist())
print("Minimum:", values.min())
print("Maximum:", values.max())
print("Range:", values.max() - values.min())
```

## 3. Variance and Standard Deviation
**Definition:** Variance measures average squared distance from the mean; standard deviation is its square root and uses the original units. Pandas uses sample variance/standard deviation by default (`ddof=1`).

**Simple example:** Scores 49, 50, 51 have less spread than 20, 50, 80.

**Business example:** Compare how consistently two stores sell each day.

**AI/ML use case:** Spread informs scaling and helps identify features with very different ranges.

**Importance:** A large standard deviation means values are more spread out, but context and units matter.

```python
for col in ["age", "monthly_salary", "monthly_spending"]:
    print(col, "variance =", round(df[col].var(), 2), "| standard deviation =", round(df[col].std(), 2))
```

## 4. Quartiles and Percentiles
**Definition:** Quartiles divide ordered data into four parts; the 25th, 50th, and 75th percentiles are Q1, median, and Q3. A percentile tells the value at or below which a given percentage of observations falls.

**Simple example:** The 90th percentile of delivery time is a threshold met by about 90% of observations.

**Business example:** A store may monitor the 95th percentile of checkout time.

**AI/ML use case:** Percentiles support outlier rules, robust scaling, and threshold selection.

**Importance:** They describe distribution position without depending only on the mean.

```python
for col in ["age", "monthly_salary", "monthly_spending"]:
    print("\n", col)
    display(df[col].quantile([0.10, 0.25, 0.50, 0.75, 0.90]))
```

## 5. Skewness and Kurtosis
**Definition:** Skewness describes distribution asymmetry. Positive skew often means a longer right tail. Kurtosis in Pandas is excess kurtosis; it describes tail heaviness relative to a normal distribution, with caveats for small samples.

**Simple example:** A few very large purchases can make spending right-skewed.

**Business example:** Revenue often has a right tail because a few customers make large purchases.

**AI/ML use case:** Strong skew can motivate log transformations for suitable positive variables.

**Importance:** Distribution shape can influence model assumptions, summaries, and outlier investigation.

```python
for col in ["age", "monthly_salary", "monthly_spending"]:
    print(col, "skewness =", round(df[col].skew(), 2), "| excess kurtosis =", round(df[col].kurt(), 2))
```

## 6. Histograms, KDE/Density, and Box Plots
**Definitions:** A histogram counts observations in bins; KDE estimates a smoothed density curve; a box plot shows median, quartiles, and potential outliers.

**Real-world example:** A clinic can inspect the distribution of patient waiting times.

**Business example:** A retailer can compare order-value spread and unusual high-value purchases.

**AI/ML use case:** Plots help decide whether to investigate outliers or transform a skewed feature.

**Importance:** Visuals reveal patterns that a single summary number can hide. KDE estimates can be unreliable for small datasets.

```python
for col in ["age", "monthly_salary", "monthly_spending"]:
    fig, axes = plt.subplots(1, 3, figsize=(15, 4))
    sns.histplot(data=df, x=col, ax=axes[0])
    axes[0].set_title(f"Histogram: {col}")
    sns.kdeplot(data=df, x=col, fill=True, ax=axes[1])
    axes[1].set_title(f"KDE: {col}")
    sns.boxplot(data=df, x=col, ax=axes[2])
    axes[2].set_title(f"Box plot: {col}")
    plt.tight_layout()
    plt.show()
```

## 7. Categorical variables: Frequency, Unique Values, and Percentage Distribution
**Definition:** Frequency is a category count; unique values are distinct labels; percentage distribution expresses each count as a share of all records.

**Simple example:** 6 Yes and 4 No responses means 60% Yes and 40% No.

**Real-world example:** Count patients by appointment type.

**Business example:** Find which product categories contribute the most orders.

**AI/ML use case:** Check whether a classification target is imbalanced and whether categories are rare.

**Importance:** Rare categories and imbalanced targets may need special handling or different metrics.

```python
for col in ["department", "education", "remote", "left_company"]:
    counts = df[col].value_counts(dropna=False)
    percent = df[col].value_counts(dropna=False, normalize=True).mul(100).round(2)
    display(pd.DataFrame({"count": counts, "percentage": percent}))
```

## 8. Bar Chart, Count Plot, and Pie Chart
**Definition:** A bar chart compares category values; a count plot shows the number of records in each category; a pie chart shows parts of a whole.

**Business example:** Compare orders across product categories.

**AI/ML use case:** Visualize target class balance or category frequency before encoding.

**Importance:** Bar/count charts are usually easier than pie charts for comparing similar-sized categories. Use pie charts sparingly.

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x="department")
plt.title("Employee count by department")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()

df["left_company"].value_counts().plot(kind="pie", autopct="%1.1f%%", figsize=(5, 5))
plt.title("Left-company distribution")
plt.ylabel("")
plt.show()
```

## 9. How to explain a finding
Example: “The salary median is the typical midpoint. If the mean is much higher than the median, a few high salaries may pull the average upward. I would inspect the histogram and box plot before deciding whether values are errors or legitimate high salaries.”

**ML pipeline impact:** Univariate analysis guides outlier checks, transformations, scaling, rare-category handling, and choice of evaluation metrics. It does not establish relationships between variables or prove causes.
