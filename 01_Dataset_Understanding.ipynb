## 1. What is a Dataset?
**Definition:** A dataset is an organized collection of related observations, usually arranged in rows and columns. Each row represents an observation and each column records one property.

**Simple example:** A table of three students with name, age, and marks is a dataset.

**Real-world example:** A hospital may keep a dataset of patient visits, symptoms, tests, and outcomes.

**Business example:** A shop stores product name, price, quantity sold, and sale date to understand performance.

**AI/ML use case:** A model learns patterns from historical examples in a dataset to predict an outcome for new cases.

**Why it matters:** If the dataset is incomplete, biased, or does not match the problem, analysis and predictions can be misleading.

## 3. What is a Feature?
**Definition:** A feature is an input variable that describes an observation. It is usually represented by a column.

**Simple example:** In a student dataset, study hours and attendance can be features.

**Real-world example:** In a house dataset, location, size, and number of bedrooms are features.

**Business example:** For sales forecasting, price, promotion, and day of week can be features.

**AI/ML use case:** Features such as age and monthly salary may help a model estimate whether an employee might leave.

**Why it matters:** Relevant features help models learn useful patterns; irrelevant or leaked features can harm performance.

## 4. What is a Target?
**Definition:** The target is the outcome a supervised machine-learning model is trained to predict.

**Simple example:** If predicting exam results, the target could be pass/fail.

**Real-world example:** A weather model may predict tomorrow's temperature.

**Business example:** A retailer may predict next month's sales.

**AI/ML use case:** `left_company` is a possible target for predicting employee attrition.

**Why it matters:** A clearly defined target determines what the model learns and how success is measured.

## 5. Independent and Dependent Variables
**Definition:** Independent variables are inputs or predictors; the dependent variable is the outcome being studied or predicted. In ML, these often correspond to features (X) and target (y).

**Simple example:** Study hours (independent) may help explain exam score (dependent).

**Real-world example:** Exercise duration may be studied in relation to heart rate, while remembering that other factors also matter.

**Business example:** Advertising spend may be compared with sales revenue.

**AI/ML use case:** Salary, department, and age can be inputs (X); employee departure can be the target (y).

**Why it matters:** Separating inputs and outcome prevents accidentally training on the answer itself. Association does not prove causation.

```python
target_column = "left_company"
identifier_columns = ["employee_id"]
feature_columns = [c for c in df.columns if c not in identifier_columns + [target_column]]
X = df[feature_columns]
y = df[target_column]
print("Feature matrix X shape:", X.shape)
print("Target y shape:", y.shape)
display(X.head())
display(y.head())
```

## 6. Numerical Features
**Definition:** Numerical features are values represented by numbers where arithmetic may be meaningful, such as salary or spending.

**Simple example:** Marks of 70, 85, and 90.

**Real-world example:** Temperature readings from sensors.

**Business example:** Revenue, cost, order quantity, and salary.

**AI/ML use case:** Models use numerical features directly or after scaling/transformations.

**Why it matters:** Numerical columns can be summarized with mean, median, spread, and plots; their units and outliers should be checked.

## 7. Categorical Features
**Definition:** Categorical features represent groups or labels, such as department or education level.

**Simple example:** Red, blue, and green are categories.

**Real-world example:** A patient's blood group is categorical.

**Business example:** Product category, payment method, or region.

**AI/ML use case:** A model may encode department or education using one-hot or ordinal encoding, depending on meaning.

**Why it matters:** Categories must be consistent; `IT`, `it`, and `IT ` may otherwise be treated as different values.

## 8. Discrete Features
**Definition:** Discrete values are countable, often whole-number counts.

**Simple example:** Number of books: 0, 1, 2, 3.

**Real-world example:** Number of visits to a clinic.

**Business example:** Units sold per order.

**AI/ML use case:** Number of previous complaints may help predict customer churn.

**Why it matters:** Discrete variables often have a limited set of possible values and may need count-aware summaries or models.

## 9. Continuous Features
**Definition:** Continuous variables can take any value within a range in principle, including fractional values.

**Simple example:** A person's height may be 170.5 cm.

**Real-world example:** Time taken to travel or sensor temperature.

**Business example:** Delivery time in hours or product weight.

**AI/ML use case:** Continuous measurements may be inputs to regression models.

**Why it matters:** Histograms, distributions, outliers, units, and scaling can affect how these features are analyzed and modeled.

## 10. Identifier Columns
**Definition:** An identifier uniquely labels a record, such as employee ID or order number. It usually names the record rather than measuring a meaningful quantity.

**Simple example:** Student ID 1024.

**Real-world example:** A tracking number for a parcel.

**Business example:** Invoice number.

**AI/ML use case:** Usually remove IDs from model inputs unless the ID carries a justified signal. IDs can encourage memorization or data leakage.

**Why it matters:** Numeric IDs should not automatically be analyzed as quantities or treated as meaningful numeric features.

## 11. Target Variable
**Definition:** The target variable is the specific dependent variable selected for the modeling objective. It can be numeric (regression) or categorical (classification).

**Simple example:** Predicting a student's final mark uses the mark as a numeric target.

**Real-world example:** Predicting whether a machine will fail uses failure status as a categorical target.

**Business example:** Predicting whether a customer will renew a subscription.

**AI/ML use case:** `left_company` can be the target for employee attrition classification.

**Why it matters:** Target labels must be accurate, available at the correct time, and consistent with the business question.

## 12. Metadata
**Definition:** Metadata is information describing the dataset, its source, structure, collection process, units, and limitations.

**Simple example:** “Marks are out of 100; collected in June.”

**Real-world example:** Sensor brand, measurement units, and collection time.

**Business example:** Sales data source, currency, region coverage, and reporting period.

**AI/ML use case:** Metadata helps assess whether training data matches production data and whether the dataset has known limitations.

**Why it matters:** Without metadata, analysts may misunderstand units, time coverage, or how values were collected.

## 13. Dataset Dimensions
**Definition:** Dataset dimensions describe the number of rows and columns, commonly shown as `(rows, columns)` in Pandas.

**Simple example:** 10 students and 4 fields gives 10 rows and 4 columns.

**Real-world example:** 1 million sensor readings and 12 fields.

**Business example:** Number of transactions and transaction attributes in a sales table.

**AI/ML use case:** Dimensions help estimate memory needs and understand the amount of training data.

**Why it matters:** Very small datasets may not generalize well; very wide datasets can need feature selection or memory optimization.
