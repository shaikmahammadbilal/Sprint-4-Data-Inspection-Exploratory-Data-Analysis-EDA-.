## 1. Load the dataset

## 2. Number of Rows and Number of Columns
**Definition:** Rows usually represent observations; columns represent attributes.

**Simple example:** A table with 50 orders and 6 fields has 50 rows and 6 columns.

**Real-world example:** A weather dataset may have one row per hourly reading.

**Business example:** Rows can represent invoices and columns can represent date, customer, and total.

**AI/ML use case:** Row count helps estimate how much training data is available; column count helps understand feature complexity.

**Importance:** Dimensions reveal whether the file loaded as expected and whether data volume may be too small or large.

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Shape:", df.shape)
```

## 3. Column Names
**Definition:** Column names label the fields in a table.

**Simple example:** `name`, `age`, `marks`.

**Real-world/business example:** `order_date`, `product_id`, `revenue` identify what each field means.

**AI/ML use case:** Clear names make feature selection and preprocessing pipelines easier to read.

**Importance:** Ambiguous or duplicated names cause confusion and bugs. Rename unclear columns before analysis.

```python
print(df.columns.tolist())
# Example of a safe rename:
# df = df.rename(columns={"old_name": "clear_new_name"})
```

## 4. Index
**Definition:** The index labels rows in a Pandas DataFrame. It is not necessarily a meaningful business ID.

**Simple example:** The default index is 0, 1, 2, ...

**Real-world example:** A time-series dataset may use dates as an index.

**Business example:** A report might index rows by store ID, if each row is one store.

**AI/ML use case:** Keep index alignment correct when joining features and targets; reset index after operations when appropriate.

**Importance:** Incorrect index alignment can pair the wrong input row with a target.

```python
print(df.index)
print("Index type:", type(df.index))
# Use reset_index(drop=True) when you want a fresh 0..n-1 index
display(df.head(3))
```

## 5. `head()`, `tail()`, and `sample()`
**Definitions:** `head(n)` shows first n rows, `tail(n)` shows last n rows, and `sample(n)` selects random rows.

**Simple example:** Preview five records before analyzing all data.

**Real-world/business example:** Inspect the first and last transactions to check whether dates and values look sensible.

**AI/ML use case:** Random samples can expose unusual records before building a pipeline.

**Importance:** A quick preview catches parsing errors, shifted columns, and suspicious values. Sampling is not a substitute for systematic checks.

```python
display(df.head())
display(df.tail(3))
display(df.sample(4, random_state=42))
```

## 6. `shape`
**Definition:** `shape` returns a tuple containing the number of rows and columns.

**Example:** `(12, 9)` means 12 rows and 9 columns.

**Importance:** It is a fast check after filtering, joining, or dropping rows/columns. Unexpected changes may indicate a mistake.

```python
print(df.shape)
print("Rows:", df.shape[0], "| Columns:", df.shape[1])
```

## 7. `info()`
**Definition:** `info()` prints a compact summary: column names, non-null counts, data types, and memory usage.

**Real-world/business example:** Check whether a date was accidentally read as text or whether a column has many missing values.

**AI/ML use case:** Identify columns needing type conversion or missing-value handling.

**Importance:** Incorrect types can break calculations or cause inappropriate preprocessing.

```python
df.info()
```

## 8. `describe()`
**Definition:** `describe()` summarizes numeric columns by default (count, mean, standard deviation, minimum, quartiles, maximum). `describe(include='all')` includes categorical summaries where possible.

**Simple example:** Quickly see the typical and extreme salary values.

**Business example:** Summarize typical order value and range.

**AI/ML use case:** Spot scale differences, possible outliers, or highly skewed values before modeling.

**Importance:** Summaries help decide which variables need further investigation; they do not explain why values occur.

```python
display(df.describe())
display(df.describe(include="all"))
```

## 9. `dtypes`
**Definition:** `dtypes` lists the stored data type of each column.

**Example:** `float64` for a number with missing values, `object` for text, `datetime64` for dates.

**AI/ML use case:** Numerical scaling, categorical encoding, and date-feature extraction depend on appropriate types.

**Importance:** Stored type and semantic meaning are not always the same: an employee ID can be stored as an integer but is an identifier.

```python
display(df.dtypes)
```

## 10. `nunique()`
**Definition:** `nunique()` counts distinct values per column; by default, it excludes missing values.

**Simple example:** `['A', 'A', 'B']` has two unique values.

**Business example:** Count distinct products or customers in a table.

**AI/ML use case:** High-cardinality categories may need special encoding; a column with only one value has little predictive use.

**Importance:** Helps detect constant columns, unexpected category variety, and identifiers.

```python
display(df.nunique())
display(df.nunique(dropna=False))
```

## 11. `value_counts()`
**Definition:** Counts how often each distinct value appears in a Series.

**Simple example:** Count the number of Yes and No responses.

**Business example:** Count orders by payment method.

**AI/ML use case:** Check class imbalance in a target such as `left_company`.

**Importance:** Rare categories and imbalanced targets can affect analysis, model training, and metric selection.

```python
display(df["department"].value_counts(dropna=False))
display(df["left_company"].value_counts(dropna=False))
display(df["left_company"].value_counts(normalize=True, dropna=False).mul(100).round(1))
```

## 12. Classify columns
Automatic types are a starting point; use domain knowledge to identify IDs and targets.

```python
numerical = df.select_dtypes(include=np.number).columns.tolist()
categorical = df.select_dtypes(include=["str", "category", "bool"]).columns.tolist()
datetime = df.select_dtypes(include=["datetime", "datetimetz"]).columns.tolist()
identifiers = ["employee_id"]
potential_targets = ["left_company"]
print("Numerical:", numerical)
print("Categorical:", categorical)
print("Date/time:", datetime)
print("Identifiers:", identifiers)
print("Potential targets:", potential_targets)
```

## 13. ML pipeline impact
Inspection is the first diagnostic stage. It informs missing-value treatment, type conversion, categorical encoding, date parsing, feature selection, and target/metric choices. Always validate that the data structure matches the dataset documentation.
