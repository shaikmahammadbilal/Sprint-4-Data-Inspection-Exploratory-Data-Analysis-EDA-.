## Part 1 – Theory Questions and Answers

### 1. What is EDA?

**Answer:** Exploratory Data Analysis (EDA) is the process of examining and understanding a dataset using summaries, statistics, and visualizations. It helps us discover patterns, errors, missing values, outliers, and relationships between variables.

### 2. Why is EDA performed before Machine Learning?

**Answer:** EDA helps us understand the data before training a model. It can reveal missing values, duplicates, incorrect data types, unusual values, class imbalance, and relationships between features. This helps us choose suitable preprocessing steps and avoid misleading model results.

### 3. What is the difference between univariate, bivariate, and multivariate analysis?

**Answer:** - **Univariate:** studies one variable, for example, the distribution of customer ages.
- **Bivariate:** studies the relationship between two variables, for example, age and salary.
- **Multivariate:** studies three or more variables together, for example, age, salary, and purchase amount.

### 4. How do you identify missing values in pandas?

**Answer:** Use `df.isnull()` to locate missing values, `df.isnull().sum()` to count them in each column, and `df.isnull().mean() * 100` to calculate the percentage of missing values.

### 5. How do you identify duplicate records?

**Answer:** Use `df.duplicated()` to identify duplicate rows and `df.duplicated().sum()` to count them. Use `df.drop_duplicates()` only after checking whether the repeated rows are truly unwanted.

### 6. What is an outlier?

**Answer:** An outlier is a value that is unusually far from most other values in a dataset. For example, if most monthly salaries are between ₹20,000 and ₹60,000, a salary of ₹5,00,000 may need investigation. It may be an error or a genuine high salary.

### 7. Explain the IQR method.

**Answer:** IQR means Interquartile Range. It measures the spread of the middle 50% of the data. Calculate Q1 (25th percentile), Q3 (75th percentile), and IQR = Q3 − Q1. Values below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR are commonly flagged as potential outliers.

### 8. What is skewness?

**Answer:** Skewness measures the asymmetry of a distribution. Positive (right) skew has a longer tail on the right; negative (left) skew has a longer tail on the left; skewness near zero suggests approximate symmetry.

### 9. What is correlation?

**Answer:** Correlation measures the strength and direction of association between two numerical variables. Pearson correlation ranges from −1 to +1. A positive value indicates variables tend to increase together; a negative value indicates one tends to decrease as the other increases; a value near zero indicates little linear association.

### 10. What is the difference between correlation and causation?

**Answer:** Correlation means two variables are associated. Causation means a change in one variable produces a change in another. Correlation alone does not prove causation because other factors may influence both variables.

### 11. What is multicollinearity?

**Answer:** Multicollinearity occurs when two or more input features in a model are strongly related to each other. It can make some model coefficients unstable and difficult to interpret. Correlation matrices and Variance Inflation Factor (VIF) are common ways to investigate it.

### 12. How do you identify class imbalance?

**Answer:** For a classification target, use `value_counts()` and `value_counts(normalize=True) * 100`. If one class has many more observations than another, the dataset may be imbalanced. The correct response depends on the problem and evaluation metric.

### 13. Why should outliers not always be removed?

**Answer:** Some outliers are genuine and important, such as unusually large transactions that may indicate fraud. Removing them automatically can discard useful information. First investigate their source, business meaning, and effect on the analysis or model.

### 14. Why is categorical analysis important?

**Answer:** Categorical analysis shows the frequency and proportion of groups, such as payment methods or product categories. It helps identify common categories, rare categories, inconsistent labels, and possible target-group patterns.

### 15. What should a complete EDA summary contain?

**Answer:** A useful summary includes dataset size and columns, data types, missing values, duplicates, invalid values, descriptive statistics, distributions, categorical frequencies, outlier findings, feature relationships, target analysis (if present), visual findings, limitations, and recommended next steps.

## Part 2 – Prepare a Sample Dataset

The following small customer dataset is created for practice. It intentionally includes a missing value, a duplicate row, and a high purchase amount so that you can practice EDA checks. These are illustrative data, not real customer records.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Customer_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 110],
    "Age": [22, 25, 28, 35, 42, 29, 31, 45, 26, 38, 38],
    "Monthly_Income": [18000, 25000, 32000, 45000, 52000, 30000, 35000, 60000, 24000, 42000, 42000],
    "Purchase_Amount": [1200, 1500, 1800, 2500, 3200, 1700, 2100, 25000, 1600, 2800, 2800],
    "City": ["Hyderabad", "Chennai", "Hyderabad", "Bengaluru", "Chennai",
             "Hyderabad", "Bengaluru", "Chennai", "Hyderabad", None, None],
    "Payment_Method": ["UPI", "Card", "UPI", "Cash", "Card", "UPI",
                       "Cash", "Card", "UPI", "Cash", "Cash"],
    "Purchased": ["Yes", "No", "Yes", "Yes", "No", "Yes",
                  "No", "Yes", "No", "Yes", "Yes"]
}

df = pd.DataFrame(data)
df
```

## Part 3 – Coding Exercises with Answers

### Exercise 1. Display the first five rows, dataset shape, column names, and data types.

```python
# Answer
print("First five rows:")
display(df.head())

print("Dataset shape (rows, columns):", df.shape)
print("Column names:", df.columns.tolist())
print("\nData types:")
print(df.dtypes)
```

### Exercise 2. Identify missing values and calculate their percentages.

```python
# Answer
missing_count = df.isnull().sum()
missing_percentage = (df.isnull().mean() * 100).round(2)

missing_summary = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percentage": missing_percentage
})
display(missing_summary)
```

### Exercise 3. Identify and count duplicate records.

```python
# Answer
print("Number of duplicate rows:", df.duplicated().sum())
display(df[df.duplicated(keep=False)])
```

### Exercise 4. Create a correlation heatmap for numerical features.

```python
# Answer
numeric_df = df.select_dtypes(include="number")
correlation_matrix = numeric_df.corr()

plt.figure(figsize=(7, 5))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

### Exercise 5. Detect potential outliers in `Purchase_Amount` using the IQR method.

```python
# Answer
column = "Purchase_Amount"

q1 = df[column].quantile(0.25)
q3 = df[column].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Potential outlier rows:")
display(outliers)
```

### Exercise 6. Analyze the distribution of a numerical feature.

```python
# Answer
feature = "Age"

print(df[feature].describe())
print("Skewness:", round(df[feature].skew(), 3))

plt.figure(figsize=(7, 4))
sns.histplot(df[feature].dropna(), kde=True, bins=8)
plt.title(f"Distribution of {feature}")
plt.xlabel(feature)
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

### Exercise 7. Perform categorical analysis on `Payment_Method`.

```python
# Answer
category_counts = df["Payment_Method"].value_counts()
category_percentages = df["Payment_Method"].value_counts(normalize=True).mul(100).round(2)

category_summary = pd.DataFrame({
    "Count": category_counts,
    "Percentage": category_percentages
})
display(category_summary)

category_counts.plot(kind="bar", figsize=(6, 4), title="Payment Method Counts")
plt.xlabel("Payment Method")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Exercise 8. Analyze the target variable `Purchased` and check for class imbalance.

```python
# Answer
target_counts = df["Purchased"].value_counts()
target_percentages = df["Purchased"].value_counts(normalize=True).mul(100).round(2)

target_summary = pd.DataFrame({
    "Count": target_counts,
    "Percentage": target_percentages
})
display(target_summary)

target_counts.plot(kind="bar", title="Target Class Distribution", figsize=(6, 4))
plt.xlabel("Purchased")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print("Compare the class percentages. A large difference may indicate class imbalance.")
```

### Exercise 9. Compare two numerical features using a scatter plot.

```python
# Answer
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x="Monthly_Income", y="Purchase_Amount", hue="Purchased")
plt.title("Monthly Income vs Purchase Amount")
plt.tight_layout()
plt.show()

print("Use the plot to inspect whether the two numerical features show a visible relationship.")
```

### Exercise 10. Generate a complete basic EDA summary.

```python
# Answer
print("========== EDA SUMMARY ==========")
print("Dataset shape:", df.shape)
print("\nColumn data types:")
print(df.dtypes)
print("\nMissing values:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
print("\nNumerical summary:")
display(df.describe())
print("\nCategorical summary:")
display(df.describe(include="object"))

print("\nPotential outliers using IQR:")
for col in df.select_dtypes(include="number").columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    count = ((df[col] < lower) | (df[col] > upper)).sum()
    print(f"{col}: {count}")

print("\nTarget distribution:")
print(df["Purchased"].value_counts(dropna=False))
print("\nEDA summary completed. Investigate each finding before changing the data.")
```
